```python
# Automated EDA Techniques
# ydata-profiling is a Python library that automatically creates a complete EDA (Exploratory Data Analysis) report for your dataset.
#Instead of manually writing many plots and analysis commands, it generates everything automatically in one report.

!pip install ydata-profiling
```

```python
# taken an default dataset from seaborn 
import seaborn as sns 
df = sns.load_dataset('tips')
```

```python
df.head()
```

```python
from ydata_profiling import ProfileReport

profile = ProfileReport(df,explorative=True)
# explorative=True tells ydata-profiling to generate a more detailed and advanced EDA report.
profile.to_file('output.html')
```
